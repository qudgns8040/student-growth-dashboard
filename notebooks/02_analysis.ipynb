```python
import pandas as pd

students = pd.read_csv("../data/students.csv")
dt_scores = pd.read_csv("../data/dt_scores.csv")
```

```python
print("===== students.head() =====")
print(students.head())

print("\n")

print("===== dt_scores.head() =====")
print(dt_scores.head())

print("\n")

print("===== students.info() =====")
students.info()

print("\n")

print("===== dt_scores.info() =====")
dt_scores.info()
```

```python
df = pd.merge(
    dt_scores,
    students,
    on="student_id"
)
```
